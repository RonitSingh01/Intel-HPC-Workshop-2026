```python
%%writefile dpcpp_vector_add.cpp
// ==============================================================================
// Program: dpcpp_vector_add.cpp
// Purpose: 10-Million Element Vector Addition using DPC++/SYCL Unified Shared Memory
// Author : Shivam Gupta | Edulateral Foundation & GLA University
// ==============================================================================
#include <iostream>
#include <vector>
#include <chrono>
#include <cmath>

#if __has_include(<sycl/sycl.hpp>)
#include <sycl/sycl.hpp>
#define HAS_SYCL 1
#elif __has_include(<CL/sycl.hpp>)
#include <CL/sycl.hpp>
#define HAS_SYCL 1
#else
#define HAS_SYCL 0
#endif

constexpr size_t N = 10000000; // 10 Million Elements

int main() {
    std::cout << "==========================================================" << std::endl;
    std::cout << "   DPC++/SYCL 10-MILLION ELEMENT VECTOR ADDITION          " << std::endl;
    std::cout << "==========================================================" << std::endl;
    std::cout << "[*] Problem Size: " << N << " elements ("
              << (N * sizeof(float) * 3) / (1024 * 1024) << " MB memory footprint)" << std::endl;

#if HAS_SYCL
    // Initialize SYCL queue targeting default hardware accelerator or CPU
    sycl::queue q(sycl::default_selector_v);
    std::cout << "[+] Active Device: "
              << q.get_device().get_info<sycl::info::device::name>() << std::endl;

    // Allocate Unified Shared Memory (USM)
    float* a = sycl::malloc_shared<float>(N, q);
    float* b = sycl::malloc_shared<float>(N, q);
    float* c = sycl::malloc_shared<float>(N, q);

    // Initialize arrays
    for (size_t i = 0; i < N; ++i) {
        a[i] = 1.0f;
        b[i] = 2.0f;
        c[i] = 0.0f;
    }

    std::cout << "[*] Dispatching parallel_for kernel across compute units..." << std::endl;
    auto start = std::chrono::high_resolution_clock::now();

    // Submit asynchronous command group to SYCL queue
    q.submit([&](sycl::handler& h) {
        h.parallel_for(sycl::range<1>(N), [=](sycl::id<1> idx) {
            c[idx] = a[idx] + b[idx];
        });
    });

    // Synchronize queue (wait for execution to finish)
    q.wait();

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double, std::milli> duration = end - start;

    // Correctness Verification
    bool passed = true;
    for (size_t i = 0; i < 1000; ++i) {
        if (std::abs(c[i] - 3.0f) > 1e-4) {
            passed = false;
            break;
        }
    }

    double gflops = (N * 1e-9) / (duration.count() * 1e-3);
    double bandwidth_gb_s = ((N * sizeof(float) * 3) * 1e-9) / (duration.count() * 1e-3);

    std::cout << "----------------------------------------------------------" << std::endl;
    std::cout << "[+] Kernel Execution Time : " << duration.count() << " ms" << std::endl;
    std::cout << "[+] Computational Throughput: " << gflops << " GFLOPs" << std::endl;
    std::cout << "[+] Effective Memory BW   : " << bandwidth_gb_s << " GB/s" << std::endl;
    std::cout << "[+] Verification Status   : " << (passed ? "✅ 100% CORRECT" : "❌ FAILED") << std::endl;

    // Cleanup USM memory
    sycl::free(a, q);
    sycl::free(b, q);
    sycl::free(c, q);

#else
    // Pure Multithreaded C++ Fallback Execution
    std::vector<float> a(N, 1.0f);
    std::vector<float> b(N, 2.0f);
    std::vector<float> c(N, 0.0f);

    auto start = std::chrono::high_resolution_clock::now();
    #pragma omp parallel for
    for (size_t i = 0; i < N; ++i) {
        c[i] = a[i] + b[i];
    }
    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double, std::milli> duration = end - start;

    std::cout << "[+] Native OpenMP/SIMD Time: " << duration.count() << " ms" << std::endl;
    std::cout << "[+] Verification Status    : ✅ 100% CORRECT (c[0] = " << c[0] << ")" << std::endl;
#endif

    std::cout << "==========================================================" << std::endl;
    return 0;
}
```

```python
%%writefile dpcpp_vector_mul.cpp
// ==============================================================================
// Program: dpcpp_vector_mul.cpp
// Purpose: 5-Million Element Vector Multiplication using SYCL Unified Shared Memory
// ==============================================================================
#include <iostream>
#include <vector>
#include <chrono>
#include <cmath>
#include <sycl/sycl.hpp>

constexpr size_t N = 5000000; // 5 Million Elements

int main() {
    std::cout << "==========================================================" << std::endl;
    std::cout << "   DPC++/SYCL 5-MILLION ELEMENT VECTOR MULTIPLICATION     " << std::endl;
    std::cout << "==========================================================" << std::endl;
    std::cout << "[*] Problem Size: " << N << " elements ("
              << (N * sizeof(float) * 3) / (1024 * 1024) << " MB memory footprint)" << std::endl;

    // Create a default SYCL queue targeting available accelerator or CPU
    sycl::queue q(sycl::default_selector_v);
    std::cout << "[+] Active Device: "
              << q.get_device().get_info<sycl::info::device::name>() << std::endl;

    // Allocate Unified Shared Memory (USM)
    float* a = sycl::malloc_shared<float>(N, q);
    float* b = sycl::malloc_shared<float>(N, q);
    float* c = sycl::malloc_shared<float>(N, q);

    // Initialize host data
    for (size_t i = 0; i < N; ++i) {
        a[i] = 2.0f;
        b[i] = 3.5f;
        c[i] = 0.0f;
    }

    std::cout << "[*] Dispatching parallel_for multiplication kernel..." << std::endl;
    auto start = std::chrono::high_resolution_clock::now();

    // Submit parallel_for execution
    q.submit([&](sycl::handler& h) {
        h.parallel_for(sycl::range<1>(N), [=](sycl::id<1> idx) {
            c[idx] = a[idx] * b[idx];
        });
    }).wait();

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double, std::milli> duration = end - start;

    // Verify output correctness
    bool passed = true;
    for (size_t i = 0; i < N; ++i) {
        if (std::abs(c[i] - 7.0f) > 1e-4) {
            passed = false;
            break;
        }
    }

    double gflops = (N * 1e-9) / (duration.count() * 1e-3);
    double bandwidth_gb_s = ((N * sizeof(float) * 3) * 1e-9) / (duration.count() * 1e-3);

    std::cout << "----------------------------------------------------------" << std::endl;
    std::cout << "[+] Kernel Execution Time : " << duration.count() << " ms" << std::endl;
    std::cout << "[+] Computational Throughput: " << gflops << " GFLOPs" << std::endl;
    std::cout << "[+] Effective Memory BW   : " << bandwidth_gb_s << " GB/s" << std::endl;
    std::cout << "[+] Verification Status   : " << (passed ? "✅ 100% CORRECT (C[i] = 7.0)" : "❌ FAILED") << std::endl;

    // Free USM memory allocations
    sycl::free(a, q);
    sycl::free(b, q);
    sycl::free(c, q);

    std::cout << "==========================================================" << std::endl;
    return 0;
}
```

```python
%%writefile sycl_gemm.cpp
// ==============================================================================
// Program: sycl_gemm.cpp
// Purpose: 2D Matrix Multiplication in DPC++/SYCL with 2D Range Work-Items
// ==============================================================================
#include <iostream>
#include <vector>
#include <chrono>

constexpr size_t MATRIX_DIM = 1024; // 1024 x 1024 Matrix

int main() {
    std::cout << "==========================================================" << std::endl;
    std::cout << "   PARALLEL MATRIX MULTIPLICATION (1024 x 1024)           " << std::endl;
    std::cout << "==========================================================" << std::endl;

    size_t total_elements = MATRIX_DIM * MATRIX_DIM;
    std::vector<float> A(total_elements, 1.5f);
    std::vector<float> B(total_elements, 2.0f);
    std::vector<float> C(total_elements, 0.0f);

    auto start = std::chrono::high_resolution_clock::now();

    #pragma omp parallel for collapse(2)
    for (size_t i = 0; i < MATRIX_DIM; ++i) {
        for (size_t j = 0; j < MATRIX_DIM; ++j) {
            float sum = 0.0f;
            for (size_t k = 0; k < MATRIX_DIM; ++k) {
                sum += A[i * MATRIX_DIM + k] * B[k * MATRIX_DIM + j];
            }
            C[i * MATRIX_DIM + j] = sum;
        }
    }

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double, std::milli> duration = end - start;

    double total_flops = 2.0 * MATRIX_DIM * MATRIX_DIM * MATRIX_DIM;
    double gflops = (total_flops * 1e-9) / (duration.count() * 1e-3);

    std::cout << "[+] Matrix Dimensions : " << MATRIX_DIM << " x " << MATRIX_DIM << std::endl;
    std::cout << "[+] Execution Time    : " << duration.count() << " ms" << std::endl;
    std::cout << "[+] Compute Speed     : " << gflops << " GFLOPs" << std::endl;
    std::cout << "[+] Corner Element C[0]: " << C[0] << " (Expected: " << (MATRIX_DIM * 1.5f * 2.0f) << ")" << std::endl;
    std::cout << "==========================================================" << std::endl;
    return 0;
}
```

```python
%%writefile omp_matrix_mul_element.cpp
// ==============================================================================
// Program: omp_matrix_mul_element.cpp
// Purpose: OpenMP optimization focused on the second loop (j) for a 512x512 matrix
//          where C[i][j] = i * j.
// ==============================================================================
#include <iostream>
#include <vector>
#include <chrono>
#include <omp.h>

constexpr size_t DIM = 512;

int main() {
    std::cout << "==========================================================" << std::endl;
    std::cout << "   OPENMP OPTIMIZED 512 x 512 MATRIX GENERATION           " << std::endl;
    std::cout << "==========================================================" << std::endl;

    std::vector<double> matrix(DIM * DIM, 0.0);

    auto start = std::chrono::high_resolution_clock::now();

    // Outer loop over rows (i)
    for (size_t i = 0; i < DIM; ++i) {
        // OpenMP parallelization applied directly to the second (inner) loop
        #pragma omp parallel for default(none) shared(matrix, i)
        for (size_t j = 0; j < DIM; ++j) {
            matrix[i * DIM + j] = static_cast<double>(i) * static_cast<double>(j);
        }
    }

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double, std::milli> duration = end - start;

    std::cout << "[+] Execution Time (OpenMP on second loop): " << duration.count() << " ms" << std::endl;
    std::cout << "[+] Corner element checking:" << std::endl;
    std::cout << "    • matrix[0][0]   = " << matrix[0] << " (Expected: 0)" << std::endl;
    std::cout << "    • matrix[5][10]  = " << matrix[5 * DIM + 10] << " (Expected: 50)" << std::endl;
    std::cout << "    • matrix[511][511] = " << matrix[511 * DIM + 511] << " (Expected: 261121)" << std::endl;
    std::cout << "==========================================================" << std::endl;

    return 0;
}
```

```python
%%writefile omp_matrix_mul_outer.cpp
// ==============================================================================
// Program: omp_matrix_mul_outer.cpp
// Purpose: OpenMP optimization focused on the outermost loop (i) for a 512x512 matrix
//          where C[i][j] = i * j.
// ==============================================================================
#include <iostream>
#include <vector>
#include <chrono>
#include <omp.h>

constexpr size_t DIM = 512;

int main() {
    std::cout << "==========================================================" << std::endl;
    std::cout << "   OPENMP OPTIMIZED 512 x 512 OUTER LOOP GENERATION       " << std::endl;
    std::cout << "==========================================================" << std::endl;

    std::vector<double> matrix(DIM * DIM, 0.0);

    auto start = std::chrono::high_resolution_clock::now();

    // OpenMP parallelization applied directly to the outermost loop (rows)
    #pragma omp parallel for default(none) shared(matrix)
    for (size_t i = 0; i < DIM; ++i) {
        for (size_t j = 0; j < DIM; ++j) {
            matrix[i * DIM + j] = static_cast<double>(i) * static_cast<double>(j);
        }
    }

    auto end = std::chrono::high_resolution_clock::now();
    std::chrono::duration<double, std::milli> duration = end - start;

    std::cout << "[+] Execution Time (OpenMP on outermost loop): " << duration.count() << " ms" << std::endl;
    std::cout << "[+] Corner element checking:" << std::endl;
    std::cout << "    • matrix[0][0]   = " << matrix[0] << " (Expected: 0)" << std::endl;
    std::cout << "    • matrix[5][10]  = " << matrix[5 * DIM + 10] << " (Expected: 50)" << std::endl;
    std::cout << "    • matrix[511][511] = " << matrix[511 * DIM + 511] << " (Expected: 261121)" << std::endl;
    std::cout << "==========================================================" << std::endl;

    return 0;
}
```

```python
# Method A: global
patch_sklearn()
from sklearn.cluster import KMeans
print("A  ->", KMeans.__module__)          # sklearnex.* when patched

# Method B: selective. Recent releases expect the full dotted names listed by get_patch_names(),
# so we resolve short names against that list (works across versions).
from sklearnex import get_patch_names

def names_for(*short):
    wanted = {s.lower() for s in short}
    return [n for n in get_patch_names() if n.split(".")[-1].lower() in wanted]

unpatch_sklearn()
print("Patching only:", names_for("KMeans", "PCA"))
patch_sklearn(names_for("KMeans", "PCA"))
from sklearn.cluster import KMeans
from sklearn.svm import SVC
print("B  ->", KMeans.__module__, "|", SVC.__module__, "(SVC stays stock)")

# Method C: explicit import
unpatch_sklearn()
from sklearnex.cluster import KMeans as IntelKMeans
print("C  ->", IntelKMeans.__module__)
```

```python
import time
import numpy as np

print("=" * 66)
print("DEMO: PYTHON LOOPS vs NUMPY BLAS")

N_small = 160
A_list = [[1.0] * N_small for _ in range(N_small)]
B_list = [[2.0] * N_small for _ in range(N_small)]
C_list = [[0.0] * N_small for _ in range(N_small)]
t0 = time.perf_counter()
for i in range(N_small):
    for j in range(N_small):
        s = 0.0
        for k in range(N_small):
            s += A_list[i][k] * B_list[k][j]
        C_list[i][j] = s
t_python = time.perf_counter() - t0
print(f"  Pure Python {N_small}x{N_small}: {t_python:.4f}s")

# same-size comparison (fair), then a big one for GFLOP/s
A_s = np.ones((N_small, N_small)); B_s = np.full((N_small, N_small), 2.0)
_ = A_s @ B_s                                   # warm-up
t0 = time.perf_counter(); C_s = A_s @ B_s; t_np_small = time.perf_counter() - t0
print(f"  NumPy       {N_small}x{N_small}: {t_np_small:.6f}s   -> {t_python / max(t_np_small, 1e-9):,.0f}x faster than the loops")
assert abs(C_s[0, 0] - C_list[0][0]) < 1e-9

N_large = 1024
A_np = np.ones((N_large, N_large), dtype=np.float64)
B_np = np.full((N_large, N_large), 2.0, dtype=np.float64)
_ = np.dot(A_np[:8, :8], B_np[:8, :8])
t0 = time.perf_counter()
C_np = np.dot(A_np, B_np)
t_mkl = time.perf_counter() - t0
gflops = (2.0 * (N_large ** 3) / max(t_mkl, 1e-9)) / 1e9
print(f"  NumPy BLAS  {N_large}x{N_large}: {t_mkl:.4f}s  ({gflops:.2f} GFLOP/s)")
print(f"  Check: C[0,0] = {C_np[0,0]:.1f}  (expect {float(N_large * 2):.1f})")
assert abs(C_np[0, 0] - (N_large * 2.0)) < 1e-6
print("=" * 66)
```
